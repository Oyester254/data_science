## 6. Clean Dataset 1

The first dataset already contains the variables needed for the
analysis, but two data-quality issues need to be addressed:

- `price_usd` is stored as text rather than a numeric value.
- Some properties have missing latitude or longitude values.

Because location is one of the factors in the research question,
records without coordinates cannot be used for location-based
analysis. These incomplete records will therefore be removed.

The cleaning process will:

1. Remove rows with missing values
2. Clean the `price_usd` values
3. Convert `price_usd` from text to numeric

````python
# Clean and standardize Dataset 1
df1 = (
    pd.read_csv(path_1)
    .dropna()
    .assign(
        price_usd=lambda x:(
            x["price_usd"]
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
            .astype(float)
        )
    )

)
````

### Cleaning the Price Variable

The `price_usd` column contains values such as:

`$125,000`

These values are stored as strings because of the dollar sign and
thousands separator.

To make the variable suitable for numerical analysis, I:

1. Remove the `$` symbol
2. Remove the comma
3. Convert the resulting string to `float`

The `regex=False` argument ensures that `$` is interpreted as a
literal character rather than a regular-expression symbol.

````python
# Verify the cleaned structure
df1.head()
````

## 7. Clean Dataset 2

Dataset 2 has the same general property information as Dataset 1,
but its price variable is stored in Mexican pesos (`price_mxn`)
instead of US dollars.

To combine the datasets consistently, prices must be converted to
the same currency.

For this project, the conversion uses:

> **19 MXN = 1 USD**

The cleaning process will:

1. Convert `price_mxn` to USD
2. Create a standardized `price_usd` column
3. Remove the original `price_mxn` column
4. Remove incomplete records

````python
df2 = (
    pd.read_csv(path_2)
    .assign(
        price_usd = lambda x: x["price_mxn"].div(19)
    )
    .drop(columns=["price_mxn"])
    .dropna()
)


````

## 8. Clean Dataset 3

Dataset 3 requires more structural transformation than the first
two datasets.

Two columns contain multiple pieces of information:

- `lat-lon` contains both latitude and longitude in one value.
- `place_with_parent_names` contains hierarchical location
  information, including the country, state, and city.

These composite fields do not follow tidy-data principles because
multiple variables are stored together.

The cleaning process will:

1. Remove incomplete records
2. Split `lat-lon` into `lat` and `lon`
3. Extract the state from `place_with_parent_names`
4. Remove the original composite columns

````python
# Clean and standardize dataset 3
df3 = (
    pd.read_csv(path_3)
    .dropna()
    .assign(
        # Extract lat from lat-lon column
        lat = lambda x: (
            x["lat-lon"]
            .str.split(",", expand=True)[0]
            .astype(float)
        ),
        # Extract lon from lat-lon column
        lon = lambda x: (
            x["lat-lon"]
            .str.split(",", expand=True)[1]
            .astype(float)
            
        ),
        # Extract state from place_with_parent_names column
        state = lambda x: (
            x["place_with_parent_names"]
            .str.split("|", expand=True)[2]
        )
        

     )
    .drop(columns=["place_with_parent_names", "lat-lon"])
)
````

### Extracting Latitude and Longitude

The original `lat-lon` column stores two variables in one cell.

For example:

`19.4326,-99.1332`

The string is split at the comma:

```python
.str.split(",", expand=True)

````python
# Preview the cleaned dataset
df3.head()
````

### Extracting the State

The `place_with_parent_names` column contains hierarchical
location information such as:

`|Mexico|Jalisco|Guadalajara|`

Splitting the string at `|` produces:

| Index | Value |
|---:|---|
| 0 | empty |
| 1 | Mexico |
| 2 | Jalisco |
| 3 | Guadalajara |

The state is therefore extracted using index `2`.

## 10. Combine the Datasets

The three cleaned datasets now have a consistent schema, so they can
be combined into a single analytical dataset.

Because each dataset contains different property listings with the
same variables, the appropriate operation is to **stack the rows**
using `pd.concat()`.

The `ignore_index=True` argument creates a new sequential index for
the combined dataset.

````python
# Combine the three cleaned datasets
df = pd.concat(
    [df1, df2, df3],
    ignore_index=True
)
````

### Why `pd.concat()`?

`pd.concat()` is appropriate because the datasets contain the same
variables but different observations.

Conceptually:

```text
Dataset 1
────────────
Property 1
Property 2
Property 3
     ↓
Dataset 2
────────────
Property 4
Property 5
Property 6
     ↓
Dataset 3
────────────
Property 7
Property 8
Property 9
